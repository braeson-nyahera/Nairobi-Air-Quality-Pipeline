```python
import requests
import os

from dotenv import load_dotenv

load_dotenv()
```

```python
BASE_URL = 'https://api.openaq.org/v3'
API_URL = 'https://api.openaq.org/v3'
API_KEY = os.getenv('OpenAQ_API_KEY')
NEWS_API_KEY = os.getenv('NEWS_API_KEY')
```

```python
def test_endpoint():
    headers = {
    "X-API-Key": API_KEY,
    }
    params = {
        "coordinates":  "-1.286389,36.817223",
        "radius": 25000,
        "limit": 100
    }

    response = requests.get(f"{BASE_URL}/locations", headers=headers, params=params)
    response.raise_for_status()

    return response.json()

    
```

```python
data = test_endpoint()
```

```python
locations = []

for record in records:
    location_data = {
        "name" : record['name'],
        "owner" : record['owner']['name'],
        "provider" : record['provider']['name'],
        "number_of_sensors" : len(record['sensors'])}

    locations.append(location_data)

locations
```

```python
def sensor_data():
    header = {
        "X-API-Key": API_KEY,
    }
    data = requests.get(f"{BASE_URL}/sensors/16243",headers=header)

    return data.json()
```

```python
sp_sensor_data = sensor_data()
sp_sensor_data
```

```python
def formatted_sensor_data(sensor_data):
    sp_data = {
        "sensor_id": sensor_data['results'][0]['id'],
        "sensor_name":sensor_data['results'][0]['name'],
        "latest_date": sensor_data['results'][0]['latest']['datetime']['local'],
        "latest_value": sensor_data['results'][0]['latest']['value'],
        "latitude": sensor_data['results'][0]['latest']['coordinates']['latitude'],
        "longitude": sensor_data['results'][0]['latest']['coordinates']['longitude']
    }

    print(sp_data)
```

```python
formatted_sensor_data(sp_sensor_data)
```

```python
def sensor_measurements():
    header = {
        "X-API-Key": API_KEY,
    }
    data = requests.get(f"{BASE_URL}/sensors/16244/measurements",headers=header)

    return data.json()['results'][-1]
```
